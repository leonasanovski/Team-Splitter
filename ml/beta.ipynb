```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

# Parameters for different shapes
params = [(2, 2), (0.5, 0.5), (1, 1)]

# Generate x values between 0 and 1
x = np.linspace(0, 1, 1000)

# Plot each distribution
plt.figure(figsize=(10, 6))

for a, b in params:
    y = beta.pdf(x, a, b)
    plt.plot(x, y, label=f'α={a}, β={b}')

plt.title("Different Beta Distributions")
plt.xlabel('x')
plt.ylabel('PDF')
plt.legend()
plt.grid(True)
plt.show()

```
